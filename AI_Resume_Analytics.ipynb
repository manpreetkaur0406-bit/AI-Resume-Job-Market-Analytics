```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split

from sklearn.preprocessing import LabelEncoder

from sklearn.ensemble import RandomForestRegressor

from sklearn.linear_model import LinearRegression

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
```

```python
salary_df.head()
```

```python
salary_df.info()
```

```python
salary_df = salary_df[
    [
        "experience_level",
        "employment_type",
        "job_title",
        "employee_residence",
        "remote_ratio",
        "company_location",
        "company_size",
        "salary_in_usd"
    ]
]
```

```python
salary_df.head()
```

```python
salary_df.isnull().sum()
```

```python
salary_df = salary_df.dropna()
```

```python
encoder = LabelEncoder()

categorical_columns = [
    "experience_level",
    "employment_type",
    "job_title",
    "employee_residence",
    "company_location",
    "company_size"
]

for col in categorical_columns:
    salary_df[col] = encoder.fit_transform(salary_df[col])
```

```python
salary_df.head()
```

```python
X = salary_df.drop("salary_in_usd", axis=1)

y = salary_df["salary_in_usd"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

```python
rf_model = RandomForestRegressor(
    n_estimators=200,
    random_state=42
)

rf_model.fit(X_train, y_train)
```

```python
predictions = rf_model.predict(X_test)
```

```python
mae = mean_absolute_error(y_test, predictions)

rmse = np.sqrt(mean_squared_error(y_test, predictions))

r2 = r2_score(y_test, predictions)

print("Mean Absolute Error:", round(mae,2))

print("Root Mean Squared Error:", round(rmse,2))

print("R² Score:", round(r2,3))
```

```python
comparison = pd.DataFrame({
    "Actual Salary": y_test.values,
    "Predicted Salary": predictions.astype(int)
})

comparison.head(10)
```

```python
import matplotlib.pyplot as plt

importance = pd.Series(
    rf_model.feature_importances_,
    index=X.columns
)

importance = importance.sort_values()

plt.figure(figsize=(10,6))

importance.plot(kind="barh")

plt.title("Feature Importance for Salary Prediction")

plt.xlabel("Importance")

plt.show()
```

```python
import joblib

joblib.dump(rf_model, "salary_prediction_model.pkl")

print("Model Saved Successfully!")
```

```python
import ast
```

```python
resume_df.columns = resume_df.columns.str.replace('\ufeff', '', regex=False)
```

```python
required_skills = [
    "python",
    "sql",
    "excel",
    "power bi",
    "tableau",
    "statistics",
    "communication",
    "machine learning"
]
```

```python
def calculate_ats(skill_text):

    try:
        skills = ast.literal_eval(skill_text)
        skills = [s.lower().strip() for s in skills]
    except:
        return 0

    matched = set(skills) & set(required_skills)

    score = (len(matched) / len(required_skills)) * 100

    return round(score,2)
```

```python
resume_df["ATS_Score"] = resume_df["skills"].apply(calculate_ats)
```

```python
resume_df = resume_df.sort_values(
    by="ATS_Score",
    ascending=False
)

resume_df["Rank"] = range(1, len(resume_df)+1)
```

```python
top10 = resume_df[
    [
        "Rank",
        "job_position_name",
        "ATS_Score",
        "skills"
    ]
].head(10)

top10
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))

plt.barh(
    top10["job_position_name"],
    top10["ATS_Score"]
)

plt.xlabel("ATS Score")
plt.ylabel("Job Position")
plt.title("Top 10 Ranked Resumes")

plt.gca().invert_yaxis()

plt.show()
```

```python
top10.to_csv("Top10_Resume_Ranking.csv", index=False)

print("Resume ranking saved successfully!")
```
